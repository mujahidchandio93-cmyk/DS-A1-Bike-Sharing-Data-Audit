### 4.5 Anomaly Audit

The anomaly audit identifies observations that are unusual relative to the observed distribution or that may violate logical relationships between variables. An anomaly is treated as a record requiring investigation rather than automatically being considered an error.

```python
# IQR-based anomaly screening for numerical variables

numeric_columns = [
    "temp", "atemp", "hum", "windspeed",
    "casual", "registered", "cnt"
]

anomaly_summary = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    n_anomalies = ((df[column] < lower_bound) |
                   (df[column] > upper_bound)).sum()

    anomaly_summary.append({
        "variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "anomaly_count": n_anomalies
    })

anomaly_summary = pd.DataFrame(anomaly_summary)

anomaly_summary
```

```python
# Logical consistency check for rental counts

count_mismatch = df["casual"] + df["registered"] != df["cnt"]

print("Rows where casual + registered != cnt:", count_mismatch.sum())

if count_mismatch.sum() == 0:
    print("Result: casual + registered equals cnt for every observation.")
else:
    print("Result: Some observations have inconsistent rental totals.")
```

```python
# Basic validity checks for rental counts

print("Negative casual counts:", (df["casual"] < 0).sum())
print("Negative registered counts:", (df["registered"] < 0).sum())
print("Negative total counts:", (df["cnt"] < 0).sum())
```

### Anomaly Audit Result

The IQR screening identified no potential anomalies for `temp`, `atemp`, `registered`, or `cnt`. It identified 2 observations for `hum`, 13 for `windspeed`, and 44 for `casual` as unusually high or low relative to the IQR-based thresholds.

These IQR flags are treated as potential distributional anomalies rather than data errors. Unusually high rental counts or weather measurements can represent legitimate days in the observed system. No negative values were found for `casual`, `registered`, or `cnt`. In addition, `casual + registered` equals `cnt` for all 731 observations, indicating internal consistency of the rental-count variables.

No observation was removed based solely on the IQR screening.

### 4.6 Leakage Audit

The leakage audit checks whether any available variables directly contain the target or are constructed from the target. The target for this analysis is `cnt`, the total number of daily rentals.

```python
# Check whether rental components reconstruct the target

reconstructed_cnt = df["casual"] + df["registered"]

print("Does casual + registered equal cnt for all rows?",
      (reconstructed_cnt == df["cnt"]).all())

print("Number of mismatches:",
      (reconstructed_cnt != df["cnt"]).sum())
```

```python
# Correlation of numeric variables with the target

numeric_df = df.select_dtypes(include="number")

correlation_with_cnt = (
    numeric_df.corr()["cnt"]
    .sort_values(ascending=False)
)

correlation_with_cnt
```

```python
# Variables excluded because they directly reconstruct the target

leakage_variables = ["casual", "registered"]

analysis_features = [
    "season",
    "yr",
    "mnth",
    "holiday",
    "weekday",
    "workingday",
    "weathersit",
    "temp",
    "atemp",
    "hum",
    "windspeed"
]

print("Target:", "cnt")
print("Excluded leakage variables:", leakage_variables)
print("Analysis features:", analysis_features)
```

### Leakage Audit Result

The audit confirmed that `casual + registered` equals `cnt` for all 731 observations, with zero mismatches. Therefore, `casual` and `registered` are direct components of the target and would create target leakage if used as predictors of `cnt`.

The correlation audit also shows a very high correlation between `registered` and `cnt` (r = 0.946) and a substantial correlation between `casual` and `cnt` (r = 0.673). These correlations are not interpreted as independent predictive evidence because the two variables directly compose the target.

For the subsequent analysis, `casual` and `registered` are excluded as predictors. The analysis feature set consists of calendar, weather, and normalized weather variables listed above. This prevents direct target leakage.

## 5. Bias, Measurement, and Claim Boundaries

### 5.1 Bias/Measurement Counterexample

A measurement limitation is that the dataset records weather variables using normalized measurements rather than direct physical units in the analysis file. In addition, daily rental demand is an aggregate system-level count and does not record individual users' reasons for renting or not renting a bicycle.

For example, two days with similar recorded temperature and humidity can have different rental totals because other factors affecting demand—such as events, traffic conditions, accessibility, or individual travel decisions—may differ but are not fully represented in the dataset. Therefore, the observed weather-demand association should not be interpreted as a complete explanation of daily rental variation.

### 5.2 Prohibited Claim

The analysis will not claim that weather conditions cause changes in bike-rental demand. The dataset is observational, so the analysis can describe associations between observed weather conditions and daily rental demand but cannot establish a causal effect of weather from these data alone.

A prohibited claim is:

> "Higher temperatures cause people to rent more bicycles."

A supported formulation is:

> "Daily bike-rental demand is associated with observed temperature and other weather conditions in the 2011–2012 Capital Bikeshare data."

```python
## 9. Main Association Model: Analysis Dataset

analysis_vars = [
    "cnt",
    "temp",
    "hum",
    "windspeed",
    "weathersit",
    "yr",
    "mnth",
    "holiday",
    "weekday",
    "workingday"
]

analysis_df = df[analysis_vars].copy()

print("Analysis dataset shape:", analysis_df.shape)
print("\nAnalysis variables:")
print(analysis_df.columns.tolist())
print("\nMissing values:")
print(analysis_df.isna().sum())
```

```python
## 9.1 Multiple Linear Regression

import statsmodels.formula.api as smf

model = smf.ols(
    formula="""
    cnt ~ temp + hum + windspeed
    + C(weathersit)
    + C(yr)
    + C(mnth)
    + C(holiday)
    + C(weekday)
    + C(workingday)
    """,
    data=analysis_df
).fit(cov_type="HC3")

print(model.summary())
```

```python
## 9.2 Revised Association Model

model_revised = smf.ols(
    formula="""
    cnt ~ temp + hum + windspeed
    + C(weathersit)
    + C(yr)
    + C(mnth)
    + C(weekday)
    + C(workingday)
    """,
    data=analysis_df
).fit(cov_type="HC3")

print(model_revised.summary())
```

```python
print(model_revised.summary())
```

```python
## 9.3 Regression Diagnostic Checks

import matplotlib.pyplot as plt
import numpy as np

fitted = model_revised.fittedvalues
residuals = model_revised.resid

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Residuals vs fitted values
axes[0].scatter(fitted, residuals, alpha=0.6)
axes[0].axhline(0, linestyle="--")
axes[0].set_xlabel("Fitted daily rentals")
axes[0].set_ylabel("Residuals")
axes[0].set_title("Residuals vs Fitted Values")

# Histogram of residuals
axes[1].hist(residuals, bins=30)
axes[1].set_xlabel("Residuals")
axes[1].set_ylabel("Frequency")
axes[1].set_title("Distribution of Residuals")

plt.tight_layout()
plt.show()

print("Mean residual:", residuals.mean())
print("Residual SD:", residuals.std())
print("Durbin-Watson:", model_revised.durbin_watson)
```

```python
from statsmodels.stats.diagnostic import het_breuschpagan

bp_test = het_breuschpagan(
    residuals,
    model_revised.model.exog
)

bp_results = pd.Series(
    bp_test,
    index=[
        "LM statistic",
        "LM p-value",
        "F statistic",
        "F p-value"
    ]
)

bp_results
```

```python
## 9.4 Residual Autocorrelation Check

from statsmodels.stats.stattools import durbin_watson
from statsmodels.stats.diagnostic import acorr_ljungbox

dw = durbin_watson(residuals)

print("Durbin-Watson statistic:", round(dw, 3))

ljung_box = acorr_ljungbox(
    residuals,
    lags=[7, 14],
    return_df=True
)

ljung_box
```

```python
## 9.5 Distribution Check for Daily Rental Demand

print("Skewness of cnt:", round(df["cnt"].skew(), 3))
print("Kurtosis of cnt:", round(df["cnt"].kurt(), 3))

plt.figure(figsize=(8, 5))
plt.hist(df["cnt"], bins=30)
plt.xlabel("Daily bike rentals (cnt)")
plt.ylabel("Frequency")
plt.title("Distribution of Daily Bike Rentals")
plt.show()
```

## 10.1 Interpretation and Claim Boundary

The audit found no missing values, duplicate rows, repeated dates, negative rental counts, or inconsistencies between `casual + registered` and `cnt`. The adjusted association model explained 82.9% of the observed variation in daily rental demand (R² = 0.829).

The results should not be interpreted causally. The data are observational and represent daily system-level aggregates from 2011–2012. In addition, the diagnostic checks detected heteroscedasticity (Breusch–Pagan p < 0.001) and residual serial correlation (Durbin–Watson = 1.069; Ljung–Box p < 0.001 at both 7 and 14 lags). Therefore, the model provides evidence of association within the observed dataset but does not establish that weather conditions cause changes in rental demand.

The findings should also not be generalized automatically to other cities, bike-sharing systems, or time periods.

```python
# Create Data Dictionary

data_dictionary_text = """# Data Dictionary

| Variable | Type | Description | Role / Notes |
|---|---|---|---|
| `instant` | Integer | Record index | Identifier; not used as an analysis predictor |
| `dteday` | Date/string | Calendar date | Date identifier; one unique date per observation |
| `season` | Integer | Season category coded 1–4 according to the dataset documentation | Calendar variable |
| `yr` | Integer | Year indicator: 0 = 2011, 1 = 2012 | Calendar variable |
| `mnth` | Integer | Month, coded 1–12 | Calendar variable |
| `holiday` | Integer | Holiday indicator | Calendar variable |
| `weekday` | Integer | Day-of-week category, coded 0–6 | Calendar variable |
| `workingday` | Integer | 1 if neither weekend nor holiday; otherwise 0 | Calendar variable |
| `weathersit` | Integer | Weather situation category | Weather variable; frozen daily data contain observed categories 1–3 |
| `temp` | Float | Normalized temperature measurement | Weather predictor |
| `atemp` | Float | Normalized feeling-temperature measurement | Weather variable; highly correlated with `temp` |
| `hum` | Float | Normalized humidity measurement | Weather predictor |
| `windspeed` | Float | Normalized wind-speed measurement | Weather predictor |
| `casual` | Integer | Count of casual users | Component of `cnt`; excluded as predictor to prevent target leakage |
| `registered` | Integer | Count of registered users | Component of `cnt`; excluded as predictor to prevent target leakage |
| `cnt` | Integer | Total number of rental bikes, including casual and registered users | Target variable |

## Notes

The frozen `day.csv` file contains 731 daily observations and 16 variables.

The README for the dataset defines additional weather-category information for `weathersit`; however, the frozen daily file used in this audit contains only categories 1, 2, and 3.

The hourly variable `hr` is not present in `day.csv` and was therefore not included in this analysis.

The weather variables are normalized measurements rather than direct physical-unit values in the analysis file. Therefore, regression coefficients for `temp`, `hum`, and `windspeed` are interpreted per unit of the normalized variable rather than directly as degrees Celsius, percentage humidity, or physical wind-speed units.
"""

(output_dir / "data_dictionary.md").write_text(
    data_dictionary_text,
    encoding="utf-8"
)

print("Data Dictionary created:", output_dir / "data_dictionary.md")
```
